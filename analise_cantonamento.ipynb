```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline


sns.set(style="whitegrid")

df = pd.read_csv("C:\\Users\\rosa2\\workspace\\cantonamento_ferroviario\\cantonamento_ferroviario_pt_leve.csv", sep=";")

colunas_interesse = ['extensao_km', 'nivel_automacao', 'modo_cantonamento', 'tipo_via']
df_filtrado = df[colunas_interesse].dropna()

print("Resumo estatístico das principais métricas:")
display(df_filtrado.describe())

correlacao = df_filtrado[['extensao_km', 'nivel_automacao']].corr()
print("\nMatriz de Correlação:")
display(correlacao)

plt.figure(figsize=(10, 5))
sns.boxplot(
    data=df_filtrado, 
    x='nivel_automacao', 
    y='extensao_km', 
    hue='nivel_automacao', 
    palette='Set2',
    legend=False            
)
plt.title('Distribuição da Extensão dos Trechos por Nível de Automação', fontsize=12, fontweight='bold')
plt.xlabel('Nível de Automação', fontsize=10)
plt.ylabel('Extensão do Trecho (km)', fontsize=10)
plt.ylim(0, 100) 
plt.tight_layout()

plt.savefig('grafico_automacao.png', dpi=300, bbox_inches='tight')

plt.show()
```
